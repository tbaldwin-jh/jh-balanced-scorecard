## Strategy Map Objectives → Dashboard Structure

The Capstone strategy map links perspectives in a cause-and-effect chain. The dashboard should reflect that hierarchy:

```
LEARNING & DEVELOPMENT          INTERNAL PROCESS           CUSTOMER              FINANCIAL
(Human/Info/Org Capital)   →   (Ops + Technology)    →   (CIF, CSI, Retention) → (ROE, ROA, Revenue)
        ↓                              ↓                        ↓                      ↓
   Emplify, Turnover            SLAs, Efficiency          CSI > 7, CIF Life      Shareholder Value
   Training, Tech Use           Ratio ≤ 52%               > 7 yrs, Channel >75%  $19MM+ NOI
```

## Suggested Dashboard Layout (aligned to Capstone + your notes)

### Main dashboard (L1 aggregates — max 7 KPIs)
1. **Total Deposits** (trend + vs. plan)
2. **Total Loans** by book (corporate / commercial / business / retail)
3. **ROA vs. peer group** (percentile rank)
4. **CSI** (current vs. target > 7)
5. **Efficiency Ratio** (current vs. ≤ 52%)
6. **Customer Retention** (current vs. 0.92)
7. **Commercial Loan SLA** (% within 60 days)

### Drill-down 1 — Top offenders / drivers
- Branch-level: deposits, loans, transaction cost, staffing
- Loan book: NPL% by category, commercial loan pipeline aging
- Process SLAs breaching target (teller > 2 min, deposit opening > 30 min)
- Segment attrition (commercial, SMB, mortgage, retail)
- Teller turnover by branch

### Drill-down 2 — Customer snapshot
- CIF life, products per customer, cross-sell %
- Channel transaction mix
- CSI survey responses (from Appendix B themes: account opening time, loan processing time, response time)
