## What the L1 monitors demonstrate

1. **Monitor** — Actual, your target, vs goal (Ahead / Behind) or vs last year; peer chip; sparkline; On track / At risk / Off track  
2. **Edit mode** — Wording, targets, and owners on tiles (where enabled)  
3. **Drill (max 3)** — KPI & gaps → trend & peers → drivers & action  
4. **Charter** — Bank vs CU labels, peers (UBPR vs FPR), never mix curves  

**Capstone starter KPIs:** ROA, NOI, deposits, loans · CSI, retention, channel usage · efficiency, commercial loan SLA, teller SLA · Emplify, teller turnover, staff turnover. Risk tiles on L1-4 include leverage / net worth, past due, LTD/LTS, IRR NII shock, and related CAMELS-aligned limits.
