## What the monitors demonstrate

1. **Monitor** — actual vs target, sparkline, On track / At risk / Off track  
2. **Edit / add objectives** — wording and goals; status updates immediately  
3. **Drill (max 3)** — KPI & gaps → trend & peers → drivers & action  
4. **Charter** — Bank vs CU labels, peers (UBPR vs FPR), never mix curves  

**Capstone starter KPIs:** ROA, NOI, deposits, loans · CSI, retention, channel usage · efficiency, commercial loan SLA, teller SLA · Emplify, teller turnover, staff turnover. CAMELS gap-fill on the combined/working monitors: leverage / net worth, past due, LTD/LTS, IRR NII shock.
